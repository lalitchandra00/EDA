```python
q1 = df_cpy.quantile(0.25)
q3 = df_cpy.quantile(0.75)
IQR = q3 - q1

lower_bound = q1 - IQR * 1.5
upper_bound = q3 + IQR * 1.5
```

```python
df_cpy = df_cpy[~((df_cpy < lower_bound) | (df_cpy > upper_bound)).any(axis=1)]
```

### Handling date column

```python
df['Date'] = df['date'].str.split('-').str[0]
df['Month'] = df['date'].str.split('-').str[1]
df['Year'] = df['date'].str.split('-').str[2]
```

```python
# .str enables to apply string operations in whole column
```

```python
df.head(2)
```

```python
df.info()
```

```python
# but the type is still object, convert it to int 
```

```python
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
```

```python
df.info()
```

```python
df.drop('date', axis = 1, inplace = True)
```

```python
df.head(2)
```

### Handling arr_time column 

```python
df['Arr_hr'] = df['arr_time'].str.split(':').str[0]
df['Arr_min'] = df['arr_time'].str.split(':').str[1]
```

```python
df['Arr_hr'] = df['Arr_hr'].astype(int)
df['Arr_min'] = df['Arr_min'].astype(int)
```

```python
df.drop('arr_time', axis = 1, inplace = True)
```

```python
df.head(2)
```

### Handling dep_time column

```python
df['Dep_hr'] = df['dep_time'].str.split(':').str[0]
df['Dep_min'] = df['dep_time'].str.split(':').str[1]
```

```python
df['Dep_hr'] = df['Dep_hr'].astype(int)
df['Dep_min'] = df['Dep_min'].astype(int)
```

```python
df.drop('dep_time',inplace = True, axis = 1)
```

```python
df.head(2)
```

### Handling time_taken column

```python
df[['Time_Taken_hr', 'Time_Taken_min']] = df['time_taken'].str.split(' ',expand = True)
```

```python
# expand = True, helps to convert list value to seperate column, if not used then list will be returned and 
# df[['Time_Taken_hr', 'Time_Taken_min']] youu have to do this individually f want to do together use expand which will convert 
# list to column 
```

```python
df['Time_Taken_hr'] = df['Time_Taken_hr'].str.replace('h','').astype(int)
```

```python
df['Time_Taken_min'] = df['Time_Taken_min'].str.replace('m','').astype(int)
```

```python
df.drop('time_taken', axis = 1, inplace = True)
```
